```python
def visualising_distribution (counter, flatlist):
  reviews_word_frequency_distribution_df = pd.DataFrame(counter.most_common(100),
                    columns=['Word', 'Frequency'])

  plt.figure(figsize=(10,10))
  sns.set_style("whitegrid")
  ax = sns.barplot(x="Word",y="Frequency", data=reviews_word_frequency_distribution_df.iloc[1:11,:])

  plt.figure(figsize = (10,10))
  wc = WordCloud(background_color = 'black', max_words = 1000,  max_font_size = 50)
  wc.generate(' '.join(flatlist))
  plt.imshow(wc)
  plt.axis('off')
```

```python
#create function for getting top emotion

def emotion_finder(df, colum):
  list_review=[]
  for sentence in df[colum]:
    list_review.append(sentence)
  #print(list_review)
  list_review_token=[]


  #Remove more than 512 token
  for sentence in list_review:
    tokenized_sentence = text_classifier.tokenizer.tokenize(sentence)
    if len(tokenized_sentence) <512:
      list_review_token.append(tokenized_sentence)
    else:
      continue

  #create top emotion for each review
  top_emotion_list=[]
  range_list = list_review_token[:10] #As it takes time to run this code, I recommend to set 100 or so.
  for i in range_list:
      emotion_labels_all = text_classifier(i, return_all_scores=True)
      all_emotion_labeles_sorted = sorted(emotion_labels_all[0], key = lambda x :x["score"], reverse = True)
      top_emotion_list.append(all_emotion_labeles_sorted[0])

  # adding top emotion in the list emotin_list
  emotion_list = []
  for i in top_emotion_list:
    emotion_list.append(i["label"])

  # counting the emotin
  count_emotion=Counter(emotion_list)
  #print(top_emotion_list[0])
  #print(list_review[0])
  #print(len(list_review)-1)

  #add review in the top emotion list
  for i in range(len(range_list)-1):
    #print(i)
    #print(top_emotion_list[i])
    #print(list_review[i])
    top_emotion_list[i]["review"]=list_review[i]

  return top_emotion_list, count_emotion
```

```python
#Create function to visualise the data
def visualise_top_emotion (emotion_data, title):
  categories=list(emotion_data.keys())
  number = list(emotion_data.values())

  #create a bar chart
  plt.bar(categories, number)
  plt.xlabel("Emotion")
  plt.ylabel("Count")
  plt.title(title)

  plt.show()
```
